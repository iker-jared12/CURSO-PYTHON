```python
def operaciones(numero):
  doble = numero *2
  triple = numero * 3
  cuadruple = numero * 4
  quintuple = numero * 5
  return doble, triple, cuadruple, quintuple

resultado = operaciones(2)
print(resultado)
```

```python
# Funcion con otra funcion
def saludar(nombre):
  return f"Hola {nombre}!"

def despedirse(nombre):
  return f"Adios {nombre}!"

def mensaje_completo(nombre):
  saludo = saludar(nombre)
  despedida = despedirse(nombre)
  return f"{saludo}\n{despedida}" #\n es un enter

print(mensaje_completo("Ana"))
```

```python
# Ejercicio 2: Ahora supongamos que tenemos una Heladeria
def pedir_sabor():
  sabores = ["vainilla", "chocolate", "fresa"]
  print("Sabor disponible", ",", sabores)
  #.lower () hace minusculas las letras de la palabra
  #.strip() elimina los espacios antes y despues de las palabras
  sabor = input("Elige un sabor: ").strip().lower()
  # miestras sabor no esta en sabores
  while sabor not in sabores:
    print("No tenemos ese sabor")
    sabor = input("Elige un sabor (nuevamente): ").strip().lower()
  return sabor

def pedir_cantidad():
  cantidad = int(input("Cuantos bolas de nieve?: "))
  while cantidad <= 0:
    cantidad = int(input("Cuantos bolas de nieve? (ingresa un valor valido): "))
  return cantidad

# Funcion madre (llama a otras funciones)
def pedir_helado():
  sabor = pedir_sabor()
  cant = pedir_cantidad()
  print(f"Pedido: {cant} bola(s) de {sabor}")
  return cant, sabor

total_pedidos = int(input("Cuantos clientes atenderas? : "))
for i in range(total_pedidos):
  pedir_helado()
```

## Funciones Lambda
Una funcion lambda es una funcion acortada para escribir una funcion de esta forma:
nombreFuncion = lambda parametros : accion

La diferencia entre funciones lambda y las funciones normales es que suelen ser un poco mas limitadas

```python
sumar = lambda x,y: x+y
print(sumar(3,10))
```

```python
descuentoLambda = lambda desc, precio: ((precio * (1- desc)), (precio * desc))
nuevoPrecio = descuentoLambda(0.4, 500)
print(f'El precio a pagar es: {nuevoPrecio[0]}, el descuento que se hizo fue de : {nuevoPrecio[1]}')
print(type(nuevoPrecio))
```

```python
# Calculadora -> suma, resta, producto, division y potencia
def suma(a, b):
  return a + b
def resta(a, b):
  return a - b
def producto(a, b):
  return a * b
def division(a, b):
   if b== 0:
    return "Error: no se puede dividir entre 0"
   return a / b
def potencia(a, b):
  return a ** b

def menu():
  print("-------MI CALCULADORA------")
  print("1. Suma")
  print("2. Resta")
  print("3. Producto")
  print("4. División")
  print("6. Salir")

while True:
  menu()
  try:
    op = int(input("Seleccione una opcion: "))

    if op < 1 or op > 6:
      print("Opcion invalida")

  except ValueError:
    print("Ingrese un numero valido")

  if op == 6:
    print("Saliendo...")
    break # sirve para salir del codigo o while o for

  try:
    a = float(input("Ingrese el primer numero: "))
    b = float(input("Ingrese el segundo numero: "))
  except ValueError:
    print("Error: Ingrese un numero valido")

  if op == 1:
    print("Resultado: ", suma(a, b))
  elif op == 2:
    print("Resultado: ", resta(a, b))
  elif op == 3:
    print("Resultado: ", producto(a, b))
  elif op == 4:
    print("Resultado: ", division(a, b))
  elif op == 5:
    print("Resultado: ", potencia(a, b))
  else:
    print("Opcion invalida")
```
