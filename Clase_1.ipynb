#Operaciones Matematicas

```python
# 4to nivel de importancia
# Suma Ordinaria
print("Suma ordinaria: ")
print(3+4)
variable1, variable2 = 20,40 #declaracion de variables multiples con comas ","
print(variable1 + variable2)

# Resta Ordinaria
print("Resta ordinaria: ")
print(2-5)
# Declaracion de Variables comun
v1=10.4
v2=15
resultado=v2-v1
print(resultado)
```

```python
# Tercer Nivel de importancia
# Multiplicacion Ordinaria
print("Multiplicacion ordinaria: ")
M1,M2=2,3.14
print(M1*M2)

# Division Ordinaria
print("------------")
print("Division ordinaria: ")
M1,M2,M3,M4,M5,M6=2,5,6,5,4,5
print(M1 / M2)
print(M3 / M4)
print(M5 / M6)

# Division de Piso
print("------------")
print("Division de Piso: ")
print(M1 // M2)
print(M3 // M4)
print(M5 // M6)
```

```python
# Potencias
print("Potencia")
print(49**1/2)
P1, P2 = 7, 2
print(P1**P2)
```

```python
# Area de circulo
pi = 3.14
r = 4
A = pi * (r ** 2)
A
```
