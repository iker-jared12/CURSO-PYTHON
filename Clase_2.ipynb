# Elif
Ahora hagamos un programa en el que preguntamos la calificacion del 0 a 100 y queremos saber su equivalencia en el sistema de evaluacion estadounidense. Para simplicidad, basemonos en esta:
*   A+ = 100
*   A = 90 - 99
*   B = 80 - 89
*   C = 70 - 79
*   D = 60 - 69
*   F < 60



```python
cal = 70
if cal == 100:
  print("A+")
elif cal >= 90:
  print("A")
elif cal >= 80:
  print("B")
elif cal >= 70:
  print("C")
elif cal >= 60:
  print("D")
else:
  print("F")
```

# Operadores logicos con Condiciones

```python
# and, or, not
pan= True
queso = True
mayonesa = True
jamon = False
print(pan and queso and mayonesa and jamon)
if not (pan and queso and mayonesa and jamon):
  print("No te puedes hacer el sandwich")
```

```python
cal = 1003
if cal >=0 and cal <= 100:
  if cal == 100:
    print("A+")
  elif cal >= 90:
    print("A")
  elif cal >= 80:
    print("B")
  elif cal >= 70:
    print("C")
  elif cal >= 60:
    print("D")
  else:
    print("F")
else:
  print("Error: la calificacion debe estar entre 0 y 100")
```

# El modulo %
operador matematico que devuelve el residuo de una división

```python
# 10 % 3 da 1, porque 10 dividido entre 3 cabe 3 veces y sobra 1
# 7 % 2 da 1, porque el 7 no es divicible de forma exacta entre 2

num = 9
if num % 3 == 0 or num % 5 == 0:
  print(f"El numero {num} es multiplo de 3 o 5")

# Saber si el numero es par o impar usando not
num = 6
if not num % 2 == 0:
  print(f"El numero {num} es impar")
else:
  print(f"El numero {num} es par")
```

# Ciclo FOR
lo llamados ciclos for nos ayuda a realizar acciones durante una cantidad determinada de veces.

```python
# imprimir los numeros del 1 al 4
for i in range(5): # Toma 5 como el numero final pero siempre incia en 0
  print(i)
```

```python
for num in range(2,11):
  print(num)
```

```python
# Tambien existe el rango (inicio, fin, paso)
for num in range(2,21,3):
  print(num)
```

```python
cadena = "Python"
for letra_por_letra in cadena:# Uso de for con caracteres
  print(letra_por_letra)
```

#While

```python
suma = 0
numeros = int(input("Ingresa el numero para sumar o 0 para salir: "))

while numeros != 0: #!= es equivalente a "distinto de ..."
  suma += numeros
  numeros = int(input("Ingresa el numero para sumar o 0 para salir: "))

print(f"La suma de los numeros es: {suma}")
```

```python
# usar el while como for
numero = 0
while numero <= 12:
  print(numero)
  numero += 1
```

```python
print("Juego: Adivina el Número")
numero_secreto = int(input("Jugador 1, escribe el numero secreto (1-20): "))

intentos = 0
max_intentos = 4

while intentos < max_intentos:
  intento = int(input("Adivina el numero: "))
  intentos += 1

  if intento == numero_secreto:
    print("¡Correcto!")
    break
  else:
    print("Incorrecto, intenta de nuevo.")
#opcional
if intentos == max_intentos and intento != numero_secreto:
  print("Se acabaron los intentos")
  print(f"el numero era {numero_secreto}")
```
