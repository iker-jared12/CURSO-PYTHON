JSON es un formato para guardar y transmitir datos. Es muy en APIs, archivos de configuraciion y almecenamiento de informacion. (se parece a un diccionario de python, pero es solo texto)


para que sirve?
- Sirve para convertir objetos de python
-> texto JSON y tambien texto JSON -> objetos de python

Esto se llama:
- Serializar: Convertir un objeto de python en texto JSON
- Deserializar: Convertir un texto JSON a un objeto usable de python

```python
# Funciones principales:
# 1) json.dums(obj): Convierte un diccionario u objeto de python en un string JSON
import json
obj = {"nombre": "Ana"
, "eda" : 25}
print(type(obj))
s = json.dumps(obj, ensure_ascii = False)
print(s)
print(type(s))
```

```python
# 2) json.loads(text_json): Convierte un string JSON en un diccionario python
restaurar = json.loads(s)
print(restaurar)
print(type(restaurar))
# Al venir de una pagina/persona/archivo desconocido debes de tener cuidado
'''
Pueden contener:
datos malinformados
Informacion inesperada
Campos faltantes
DAtos demasiados grandes
Va incorrectas que rompen el porgrama
'''
```

## Notas importantes:

```python
#pip install numpy
import numpy as np
A = np.array([[1,2,3],[4,5,6]])#array (2,3)
v = np.array([10,20,30]) #array (1,3)
print(A + v) #suma vectores
print("----------------------")
prueba = [1,2,3]
prueba2 = [1,2,3]
print(prueba + prueba2) #une los vectores
```

```python
# Mas funciones de Numpy
z = np.zeros((3,3)) # array (3,3)
print(z)
o = np.ones((2,4))
print(o)
r = np.random.rand(3,3)
print(r)
```

```python
# Rangos numericos
nums = np.arange(0, 10, 2) # saca los valores entre 0 y 10 de 2 en 2
print(nums)
x = np.linspace(0,1,5) # saca 5 valores espaciados de manera equitativa entre 0 y 1
print(x)
```

```python
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])
print(a +b)
print(a - b)
print(a * b)
print(a / b)
print(a ** b)
```

```python
print(a.dot(b)) # producto entre 2 matrices
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[10,20],[30,40],[50,60]])
print(c.dot(d))
```

```python
data = np.array([1,2,3,4,5])
print(data.max()) # maximo
print(data.min()) # minimo
print(data.sum()) # suma
print(data.mean()) # promedio/media
print(data.std()) # Desviacion estandar
```
