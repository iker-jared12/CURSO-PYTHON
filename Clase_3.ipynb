# Estructuras de Datos
Las estructuras de datos son formas de organizar informacion para trabajar con ellas de manera eficiente

- Listas -> ordenadas, modificadas, permiten duplicados
- Tuplas -> ordenadas, inmutables, permiten duplicados
- Setes -> no ordenadas, sin duplicados
- Diccionarios -> pares clave: valor

Nota:
- Pilas (FILO -> First in, Last out)
- Colas (FIFO -> First in, First out)
- Listas enlazadas (doblemente enlazadas)

# Listas
Las listas son colecciones ordenadas que podemos modificar y almacenan elementos de distintos tipos de datos


```python
lista_compras = ["manzanas", "pan", "leche", "mantequilla", "queso", "jamón"]
print(lista_compras[5])
```

```python
# Agregar un elemento a la lista .apped()
lista_compras.append("sopa")
print(lista_compras)
```

```python
lista_compras.insert(1, "platano")
lista_compras
```

```python
print(lista_compras)
```

```python
# Para eliminar un elemento de una lista existe el remove y el pop
lista_compras.remove("platano")
print(lista_compras)
```

```python
lista_compras.pop() # Sin nada dentro de los parentesis borra el ultimo elemento
lista_compras.pop(3) # Borra en la posicion indicada (recordando que es inciso 0)
# nota: Te retorna el numero eliminado
```

```python
# Podemos recorrer el arreglo por medio de un for
va = "Python"
va[3]
```

```python
for elementos in lista_compras:
  print(elementos)
```

```python
# Len saca el tamaño de la lista
print(len(lista_compras))
for i in range(len(lista_compras)):
  print(f"{i+1} - {lista_compras[i]}")
```

```python
promedio = [55.6, 97.5, 23.6, 97.5, 72.4, 100]
print(promedio)
promedio.sort() # Sort sirve para ordenar la lista de manera ascendente
print(promedio)

# ordenar una lista de manera descendente
promedio.sort(reverse=True)
print(promedio)
```

```python
# Podemos ingresar al ultimo elemento de una lista
print(promedio[-1])
```

```python
# Encontrar el mayor y menor de una lista
print(max(promedio))
print(min(promedio))
```

```python
# declaracion y uso de matrices
calificaciones = [[100, 88, 93], [99, 56, 72]]
print(calificaciones)
i = 1
for cali_i in calificaciones:
  print(f"\nAlumno {i}")
  j = 1
  for cali_j in cali_i:
    print(f"Cali {j}: {cali_j}")
    j += 1
  j = 0
  i += 1
```

# SETS
Los SETS son colecciones sin orden y no permiten elementos duplicados

```python
nombre = {"Angel", "Eliud", "Diana", "Juan", "Juan"}
print(nombre)
print(type(nombre))
```

```python
# Add
nombre.add("Leonardo")
nombre
```

```python
print(nombre[2])
```

```python
# Union
matutino = {20322, 22183, 123123, 53234}
vespertino = {20322, 11111, 222222, 333333}

# Union con .union
total_matriculas = matutino.union(vespertino)
print(total_matriculas)

# Union con |
total_matriculas = matutino | vespertino
print(total_matriculas)
```

```python
# i¿Interseccion
ca = {1234, 121, 646, 12414, 73434, 124124, 675234}
al = {1234, 111, 222, 333, 444}

# Interseccion con .intersection()
ca_al = ca.intersection(al)
print(ca_al)

# Interseccion usando &
ca_al = ca & al
print(ca_al)
```

```python
# Diferencia
# Diferencia usando .difference
solo_calculo = ca.difference(al)
print(solo_calculo)

# Diferencia usando -
solo_calculo = ca - al
print(solo_calculo)
```

```python
nombre = {"Angel", "Eliud", "Diana", "Juan", "Juan"}
# remove, pop
nombre.remove("Angel") # Debes eliminar el que tu quieras
print(nombre)
nombre = {"Angel", "Eliud", "Diana", "Juan", "Juan"}
nombre.pop() # Elimina un elemento de manera aleatoria
print(nombre)
```

```python
# Eliminar todo el SET/ sirve para vaciar la variable
nombre.clear()
nombre
```

#Diccionarios
Los diccionarios son colecciones no ordenadas de pares clave: valor

```python
alumno = {
    #clave    valor
    "nombre": "Eliud Sanchez",
    "edad": 21,
    "carrera": "LCC",
    "semestre": 7,
    "promedio": 88,
    "calificaciones": [100, 94, 94, 97, 100]
}
print(type(alumno))
alumno
```

```python
print("El alumno ", alumno["nombre"], "esta esdudiando ", alumno["carrera"])
```

```python
# Agregar al diccionario
alumno["area preferida"] = "Ciencia de datos"
```

```python
alumno
```

```python
#acceder al diccionario
for valor in alumno:
  print(alumno[valor])
```

```python
for aspectos, valores in alumno.items():
  # devuelve una vista de todos los pares clave: valor del diccionario
  print(f"{aspectos}: {valores}")
```

```python
# El metodo .clear() tambien se puede usar en diccionario
# Tenemos el .get que nos ayuda a obtener el valor (value) de cierta clave (key)
print(alumno.get("nombre"))
```

```python
alumno_1 = alumno.items() # .items sirve para mostrar toda la iformacion de los diccionarios
print(alumno_1)
print(type(alumno_1))

alumno_lista = list(alumno_1) # para convertir en lista
print(alumno_lista)
print(type(alumno_lista))
```

```python
alumno_1 = alumno.keys() # .keys solo trae las variables o titulos de los que se esta trabjando en el diccionario
print(alumno_1)
print(type(alumno_1))
```

```python
# El metodo .values() sirve para obtener los valores del diccionario, tambien se puede usar en iteracciones
alumno_v = alumno.values()
print(type(alumno_v))
print(alumno_v)
```

```python
if "carrera" in alumno:
  print("Existe carrera")
else:
  print("No existe carrera")

if "dinero" in alumno:
  print("Alumno tiene dinero")
else:
  print("Alumno no tiene dinero")
```

```python
# .copy() puedes copiar todo el diccionario en una nueva variable
# update -> Sirve para cambiar los valores (value) (.update())
alumno.update({"semestre": 6})
alumno
```

```python
# Lista de productos
productos = ["manzana", "banana", "leche", "pan"]  # <-- LISTA

# Tupla de precios
precios = (10, 5, 20, 15)  # <-- TUPLA

# Diccionario de inventario: producto -> {"precio": x, "stock": y}
inventario = {
    productos[i]: {"precio": precios[i], "stock": 10}
    for i in range(len(productos))
}  # <-- DICCIONARIO

# Set de productos en oferta
ofertas = {"manzana", "pan"}  # <-- SET

# Mostrar productos
for p in productos:
    msg = "OFERTA!!!" if p in ofertas else "Grandes precios!"
    info = inventario[p]
    print(f"{p.title()} - ${info['precio']} - Stock: {info['stock']} -> {msg}")

# Compra simple
compra = input("\n¿Qué producto quieres comprar? ").strip().lower()

if compra in inventario and inventario[compra]["stock"] > 0:
    inventario[compra]["stock"] -= 1
    print(f"Compraste 1 {compra}. Stock restante: {inventario[compra]['stock']}")
else:
    print("Producto no disponible.")
```
